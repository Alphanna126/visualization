```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
#scatter plot
np.random.seed(42)
x = np.arange(50)
y = np.random.randint(0, 100,50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
#bar plot

np.random.seed(42)
x = np.arange(50)
y = np.random.randint(0, 100,50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
# plot

np.random.seed(42)
x = np.arange(50)
y = np.random.randint(0, 100,50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
# histogram plot

np.random.seed(42)
x = np.arange(50)
y = np.random.randint(0, 100,50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
#add title and x,y label
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)

ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

```python
#add fonts and location

font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)

ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start',fontdict = font2)
```

```python
#activity
# library & dataset
import pandas as pd
import matplotlib.pyplot as plt

time = [17, 25, 42, 35, 26, 27, 20, 11, 22, 32, 35, 30, 45, 55]

# Create 2 columns: one categorical and one numerical
sample_size = 1000
data = {
    'type': ['group1'] * sample_size + ['group2'] * sample_size,
    'value': np.concatenate([np.random.normal(0, 1, sample_size),
                             np.random.normal(4, 1, sample_size)])
}
df = pd.DataFrame(data)

# Plot the histograms of each group
for group in df['type'].unique():
    
    # Filter the dataset on the group
    filtered_df = df[df['type']==group]
    
    # Add the histogram to the graphic
    filtered_df['value'].hist(figsize=(8, 4))

# Display the plot    
plt.show()
```
