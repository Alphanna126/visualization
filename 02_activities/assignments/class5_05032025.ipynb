```python
#import packages
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
#import dataset
tips = sns.load_dataset("tips")
print(tips)
```

```python
#simple line plot of tip amount vs total bill with style as whitegrid

sns.set_style('whitegrid')

tipgraph = sns.lineplot(data=tips, # choose our dataset
x='total_bill', # define our x variable
y='tip')

#set title and x,y label
tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')
```

```python
#set figure size
fig = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip',
color = 'hotpink',
linestyle = '--',
linewidth = 3,
marker = 'o',
markerfacecolor = 'indigo')

tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')
14
```

```python
#add multiple variable

tipgraph = sns.scatterplot(data=tips, x='total_bill',
y='tip', style = 'time', hue =
'day', palette = ['purple',
'hotpink', 'deepskyblue',
'yellowgreen'])

tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')
```

```python
#pairplot
sns.pairplot( data = tips, hue = 'day')
```

```python
#activity
daysplot = sns.relplot(
data=tips, # refers to data used
x="total_bill", #x-label
y="tip", #y-label
hue="sex", #column chosen for color based on sex variable
col="day", # column chosed for comparison
kind="scatter", #define what kind of plot is to be used
col_wrap=2) #show in 2 columns and 2 rows(i.e columns in a row)
```

```python
import plotly.graph_objects as go # 'go' is 'graph objects’
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
#make our figure
graph = go.Figure()
#choose our plot type (bar) and our data
graph.add_trace(go.Bar(x=x1, y=y1))
```

```python
#update our layout to include titles
graph.update_layout(
title="Pirate Scores",
xaxis_title="Pirates",
yaxis_title="Score")

graph.show()
```

```python
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', # we want points for a scatter plot
marker=dict(
size=15, # point size
color='hotpink', # point colour
opacity=1, # point transparency/alpha
line=dict(width=5, color='purple') # point outline
)))

graph.update_layout(
title='Interactive Pirate Plot',
xaxis_title='Pirates',
yaxis_title='Scores',
width=500, height=500)
```

```python
#!pip install wordcloud
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
on_bad_lines='skip')
df
```

```python
#join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote)
# generate our wordcloud image
wordcloud = WordCloud(background_color="white",
colormap = 'inferno').generate(text)
# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud, # remember 'imshow' from when we added pictures to our matplotlib axes
interpolation='bilinear') # this line helps smooth our image
ax.axis("off")
plt.show()
```

```python
#First import the package
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

#Then define our sets
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python
diagram = venn2_unweighted([A, B],
set_labels = ('Fruits', 'Pies'),
set_colors=("blue", "red"),
alpha=0.5) #transparency
plt.show()
```

```python
diagram = venn2([A, B],
set_labels = ('Fruits', 'Pies'),
set_colors=("blue", "red"),
alpha=0.5) #transparency
diagram.get_label_by_id("10").set_text("\n".join(A - B)) #left
diagram.get_label_by_id("11").set_text("\n".join(A & B)) #middle
diagram.get_label_by_id("01").set_text("\n".join(B - A)) #right
#10,11,01 are numbers for positioning
```
